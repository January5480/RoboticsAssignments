### 自定义层

```python
import torch
import torch.nn.functional as F
from torch import nn 
```

不带参数的层

```python
class CenteredLayer(nn.Module):
    def __init__(self):
        super().__init__()
    def forward(self, Y):
        return Y - Y.mean()
```

```python
layer = CenteredLayer()
layer(torch.FloatTensor([1,2,3,45,3]))
```

```python
net = nn.Sequential(nn.Linear(6, 128), layer)
Y = net(torch.randn(3,6))
Y.sum()
```

带参数的层

```python
class MyLinear(nn.Module):
    def __init__(self, in_units, units):
        super().__init__()
        self.weight = torch.randn(in_units, units)
        self.bias = torch.randn(units)
    def forward(self, X):
        linear = torch.matmul(X, self.weight) + self.bias
        return F.relu(linear) 
        
```

```python
linear = MyLinear(5,3)
linear.weight
```

```python
linear(torch.rand(2,5))
```

```python
net = nn.Sequential(MyLinear(64, 8), MyLinear(8, 1))
net(torch.rand(3, 64))
```

设计一个接受输入并计算张量降维的层torch.squeeze（）

```python
class Squee(nn.Module):
    def __init__(self, W):
        super().__init__()
        self.wi = W.shape[0]
        self.Wj = W.shape[1]
        self.Wk = W.shape[2]
        self.W = W
    def forward(self, X):
        y = torch.randn(self.Wk,1)
        for i in range(self.Wk):
            y[i] = torch.sum(torch.mul(torch.matmul(X, X.T), self.W[:, :, i]))
        return y
```

```python
W = torch.randn(size=(3,3,3))
sq = Squee(W)
X = torch.randn(3,1)
sq(X)
```
